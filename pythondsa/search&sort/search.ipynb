```python
# The list must be sorted for binary search to work
l = [23, 56, 67, 78, 89, 90]
a = 90

low = 0
high = len(l) - 1
found = False

while low <= high:
    mid = (low + high) // 2  # Recalculate mid inside the loop
    
    if l[mid] == a:
        print("element is found :",mid+1)
        found = True
        break
    elif l[mid] < a:
        low = mid + 1        # Search the right half
    else:
        high = mid - 1       # Search the left half (Corrected from low=mid-1)

if not found:
    print("element not found:")
```

```python
def bubblesort(a):
    n=len(a)

    for i in range(n):
        for j in range(n-i-1):
            if (a[j] > a[j+1]):
                a[j],a[j+1]=a[j+1],a[j]
    return a
 
a=[78,98,23,19,45,9,87,10]
print(bubblesort(a))
```

```python
def selectionsort(a):
    n=len(a)
    for i in range(n):
        min_idx = i
        for j in range(i+1,n):
            if(a[min_idx] > a[j]):
                min_idx = j
        
        a[i],a[min_idx] = a[min_idx],a[i]
    return a



a=[78,98,23,19,45,9,87,10]
selectionsort(a)
print(a)
```

```python
def insertionsort(a):
    n=len(a)
    for i in range(1,n):
        key =a[i]
        j=i-1
        while(j>=0 and key<a[j]):
            a[j+1]=a[j]
            j=j-1
        a[j+1]= key

a=[78,98,23,19,45,9,87,10]
insertionsort(a)
print(a)
```
