```python
arr=[[2,3,4],[10,8,9],[5,11,15]]
for i in arr:
    for j in i:
        print(j,end=" ")
    print()
```

```python
print(arr[1])
print(arr[2][0])
```

```python
arr.insert(2,[7,9,10])
print("modified array")
for i in arr:
    for j in i:
        print(j,end=" ")
    print()
```

```python
arr[2].insert(2,12) #arr[row number].insert(column number,element)
for i in arr:
    for j in i:
        print(j,end=" ")
    print()
```

```python
arr[1]=[8,4,0]
for i in arr:
    for j in i:
        print(j,end=" ")
    print()
```

```python
arr[1][2]=9
for i in arr:
    for j in i:
        print(j,end=" ")
    print()
```
