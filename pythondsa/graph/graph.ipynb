```python
#GRAPH-MATRIX REPRESENTATION

class graph:
    def __init__(self,vertex):
        self.mat=[[0]*vertex for x in range(vertex)]
        self.size=vertex
    #undirected
    def add_edge(self,src,dest):
        if (0<= src <self.size and 0<=dest<self.size):
            self.mat[src][dest]=1
            self.mat[dest][src]=1
        else:
            print("invalid edge")
    #directed
    def add_edgee(self,src,dest):
        if (0<= src <self.size and 0<=dest<self.size):
            self.mat[src][dest]=1
        else:
            print("invalid edge")
    #undirected-weighted
    def add_edgeee(self,src,dest,weight):
        if (0<= src <self.size and 0<=dest<self.size):
            self.mat[src][dest]=weight
            self.mat[dest][src]=weight
        else:
            print("invalid edge")

    #directed-weighted
    def add_edgeeee(self,src,dest,weight):
        if (0<= src <self.size and 0<=dest<self.size):
            self.mat[src][dest]=weight
        else:
            print("invalid edge")
    #print edges
    def print(self):
        for row in self.mat:
            print(' '.join(map(str,row)))

    

    

    
g=graph(5)
g.add_edge(0,1)
g.add_edge(0,2)
g.add_edge(1,3)
g.add_edge(3,4)
g.add_edge(2,3)
g.add_edge(2,4)
g.print()
print()

g=graph(5)
g.add_edgee(0,1)
g.add_edgee(0,2)
g.add_edgee(1,3)
g.add_edgee(3,4)
g.add_edgee(2,3)
g.add_edgee(2,4)
#g.print()
print()

g=graph(5)
g.add_edgeee(0,1,9)
g.add_edgeee(0,2,3)
g.add_edgeee(1,3,7)
g.add_edgeee(3,4,2)
g.add_edgeee(2,3,1)
g.add_edgeee(2,4,5)
#g.print()
print()

g=graph(5)
g.add_edgeeee(0,1,9)
g.add_edgeeee(0,2,3)
g.add_edgeeee(1,3,7)
g.add_edgeeee(3,4,2)
g.add_edgeeee(2,3,1)
g.add_edgeeee(2,4,5)
#g.print()




```

```python
#GRAPH-ADJACENCY LIST
class graph:
    def __init__(self):
        self.adjlist ={}
    def add_vertex(self,vertex):
        if vertex not in self.adjlist:
            self.adjlist[vertex]=[]
    def addedge(self,src,dest):
        self.add_vertex(src)
        self.add_vertex(dest)

        self.adjlist[src].append(dest)
        self.adjlist[dest].append(src)
    def printgraph(self):
        for vertex in self.adjlist:
            print(vertex, "->",self.adjlist[vertex] ,end='\n')

g=graph()
g.addedge(1,2)
g.addedge(1,4)
g.addedge(2,3)
g.addedge(2,4)
g.addedge(4,5)
g.addedge(4,3)
g.addedge(3,5)

g.printgraph()
```

```python
from collections import deque
class graph:
    def __init__(self,vertex):
        self.mat=[[0]*vertex for x in range(vertex)]
        self.size=vertex
    #undirected
    def add_edge(self,src,dest):
        if (0<= src <self.size and 0<=dest<self.size):
            self.mat[src][dest]=1
            self.mat[dest][src]=1
        else:
            print("invalid edge")

    def BFS(self,src):
        visited = [False]* self.size
        queue= deque([src])
        visited[src]= True

        while(queue):
            v= queue.popleft()
            print(v ,end=' ')
            
            for i in range(self.size):
                if(self.mat[v][i]== 1 and visited[i] == False):
                    visited[i]=True
                    queue.append(i)

    

g2=graph(8)
g2.add_edge(0,1)
g2.add_edge(0,3)
g2.add_edge(1,3)
g2.add_edge(3,5)
g2.add_edge(3,4)
g2.add_edge(4,5)
g2.add_edge(4,6)
g2.add_edge(6,2)
g2.add_edge(6,7)
g2.BFS(0)

```
