```python
class queue:
    def __init__(self):
        self.items=[]

    def isempty(self):
        return len(self.items) == 0
    def insert(self,value):
        self.items.append(value)
    
    def delete(self):
        if(self.isempty()):
            print("queue is empty")
        else:
            return (self.items.pop(0))
    
q1= queue()
q1.insert(20)
q1.insert(90)
q1.insert(100)

print(q1.delete())
print(q1.delete())
print(q1.delete())
print(q1.delete())



```

```python
#double ended queue is known as  deque
class deque:
    def __init__(self):
        self.items=[]

    def isempty(self):
        return len(self.items) == 0
    
    def insertatend(self,value):
        self.items.append(value)
    
    def deleteatfront(self):
        if(self.isempty()):
            print("queue is empty")
        else:
            return (self.items.pop(0))
    
    def insertatfront(self,value):
        return self.items.insert(0,value)

    def deleteatend(self):
        if(self.isempty()):
            print("queue is empty")
        else:
            return self.items.pop()

dq= deque()
dq.insertatend(10)
dq.insertatend(20)
dq.insertatfront(30)

print(dq.deleteatend())
print(dq.deleteatfront())
```
