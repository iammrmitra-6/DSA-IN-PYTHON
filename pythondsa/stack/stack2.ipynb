```python
##stack program using class
class stack:
    def __init__(self):
        self.s=[]

    def isempty(self):
        if len(self.s)==0:
            return False
        else:
            return True
    def push(self,items):
        self.s.append(items)

    def pop(self):
        if len(self.s)==0:
            return 'underflow'
        else:
            val=self.s.pop()
        return val
    def peek(self):
        if len(self.s)==0:
            return 'underflow'
        else:
            return self.s[len(self.s)-1]
        
    def show(self):
        if len(self.s)==0:
            return 'underflow'
        else:
            t=len(self.s)-1
            while(t>=0):
                print(self.s[t],'<==',end='')
                t-=1
            print()

        
stk=stack()
print(stk.isempty())
stk.push(10)
stk.push(20)
stk.push(40)
stk.push(60)
stk.show()
print(stk.isempty())

print(stk.peek())
print(stk.pop())

stk.show()

    
```
