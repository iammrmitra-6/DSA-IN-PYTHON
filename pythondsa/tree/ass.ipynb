```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def insert(root, key):
    if root is None:
        return Node(key)
    if key < root.data:
        root.left = insert(root.left, key)
    elif key > root.data:
        root.right = insert(root.right, key)
    return root


def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)


root = None
elements = [50, 30, 70, 20, 40]
for el in elements:
    root = insert(root, el)

print("BST created successfully with elements:", elements)
print("Inorder Traversal of BST:")
inorder(root)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def insert(root, key):
    if root is None:
        return Node(key)
    if key < root.data:
        root.left = insert(root.left, key)
    elif key > root.data:
        root.right = insert(root.right, key)
    return root


def preorder(root):
    if root is not None:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)


root = None
items = [50, 30, 20, 40, 70, 60, 80]
for item in items:
    root = insert(root, item)

print("Preorder Traversal:")
preorder(root)
```
