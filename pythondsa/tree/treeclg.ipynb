```python
# General Tree

class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []


# Search a node
def search(root, value):
    if root.data == value:
        return root

    for child in root.children:
        result = search(child, value)
        if result:
            return result

    return None


# Insert a new node
def insert(root, parent_value, new_value):
    parent = search(root, parent_value)

    if parent:
        parent.children.append(TreeNode(new_value))
        print("Node inserted successfully.")
    else:
        print("Parent node not found.")


# Delete a node
def delete(root, value):
    for child in root.children:
        if child.data == value:
            root.children.remove(child)
            print("Node deleted successfully.")
            return True

        if delete(child, value):
            return True

    return False


# Update a node
def update(root, old_value, new_value):
    node = search(root, old_value)

    if node:
        node.data = new_value
        print("Node updated successfully.")
    else:
        print("Node not found.")


# Display Tree
def display(root, level=0):
    print(" " * level * 4 + str(root.data))

    for child in root.children:
        display(child, level + 1)


# ---------------- Main Program ----------------

root_value = input("Enter root node: ")
root = TreeNode(root_value)

while True:
    print("\n1. Insert")
    print("2. Delete")
    print("3. Update")
    print("4. Display")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        parent = input("Enter parent node: ")
        new = input("Enter new node: ")
        insert(root, parent, new)

    elif choice == 2:
        value = input("Enter node to delete: ")

        if root.data == value:
            print("Root node cannot be deleted.")
        else:
            if not delete(root, value):
                print("Node not found.")

    elif choice == 3:
        old = input("Enter node to update: ")
        new = input("Enter new value: ")
        update(root, old, new)

    elif choice == 4:
        print("\nTree Structure:")
        display(root)

    elif choice == 5:
        break

    else:
        print("Invalid choice.")
```

```python
# Binary Search Tree (BST)

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Insert
def insert(root, data):
    if root is None:
        return Node(data)

    if data < root.data:
        root.left = insert(root.left, data)

    elif data > root.data:
        root.right = insert(root.right, data)

    return root


# Search
def search(root, key):
    if root is None:
        return False

    if root.data == key:
        return True

    if key < root.data:
        return search(root.left, key)
    else:
        return search(root.right, key)


# Find Minimum
def minimum(root):
    while root.left:
        root = root.left

    return root


# Delete
def delete(root, key):
    if root is None:
        return root

    if key < root.data:
        root.left = delete(root.left, key)

    elif key > root.data:
        root.right = delete(root.right, key)

    else:
        # Node with one child
        if root.left is None:
            return root.right

        elif root.right is None:
            return root.left

        # Node with two children
        temp = minimum(root.right)
        root.data = temp.data
        root.right = delete(root.right, temp.data)

    return root


# Update
def update(root, old, new):
    if search(root, old):
        root = delete(root, old)
        root = insert(root, new)
        print("Node Updated Successfully.")
    else:
        print("Node Not Found.")

    return root


# Inorder
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data, end=" ")
        inorder(root.right)


# Preorder
def preorder(root):
    if root:
        print(root.data, end=" ")
        preorder(root.left)
        preorder(root.right)


# Postorder
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data, end=" ")


# ---------------- MAIN PROGRAM ----------------

root = None

while True:
    print("\n===== Binary Search Tree Menu =====")
    print("1. Create Tree")
    print("2. Insert Node")
    print("3. Search Node")
    print("4. Update Node")
    print("5. Delete Node")
    print("6. Inorder Traversal")
    print("7. Preorder Traversal")
    print("8. Postorder Traversal")
    print("9. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        value = int(input("Enter Root Value: "))
        root = Node(value)
        print("BST Created Successfully.")

    elif choice == 2:
        value = int(input("Enter Value to Insert: "))
        root = insert(root, value)
        print("Node Inserted.")

    elif choice == 3:
        value = int(input("Enter Value to Search: "))

        if search(root, value):
            print("Element Found.")
        else:
            print("Element Not Found.")

    elif choice == 4:
        old = int(input("Enter Old Value: "))
        new = int(input("Enter New Value: "))
        root = update(root, old, new)

    elif choice == 5:
        value = int(input("Enter Value to Delete: "))
        root = delete(root, value)
        print("Node Deleted.")

    elif choice == 6:
        print("Inorder Traversal:")
        inorder(root)
        print()

    elif choice == 7:
        print("Preorder Traversal:")
        preorder(root)
        print()

    elif choice == 8:
        print("Postorder Traversal:")
        postorder(root)
        print()

    elif choice == 9:
        print("Thank You!")
        break

    else:
        print("Invalid Choice.")
```
